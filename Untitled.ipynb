```python
# Подключение библиотек

import pandas as pd
import numpy as np
import requests
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
# Загрузка и фильтрация датафрейма. Компьютеру было плохо от всего датафрейма, я взял первые 10000 строк

link = 'data/registry.xlsx'
df = pd.read_excel(link, skiprows=2, nrows=10000)
df = df[df['Основной вид деятельности'] == '41.20 Строительство жилых и нежилых зданий']

values = ['Среднее предприятие', 'Малое предприятие']
df = df[df['Категория'].isin(values)]
df = df[df['Тип субъекта'] == 'Юридическое лицо']
df.reset_index(inplace=True, drop=True)

df['id'] = np.nan
```

```python
# Поиск ID страницы предприятия по ИНН

session = requests.Session()
headers = {"User-Agent": "...Chrome..."}

n = len(df)
for i in range(n):
    inn = df['ИНН'][i]
    url = f'https://bo.nalog.gov.ru/advanced-search/organizations?inn={inn}&hasKfo=false&kfoSearch=false&page=0&size=20'
    temp = session.get(url, headers = headers).json()['content']
    if len(temp) > 0:
        use_id = temp[0]['id']
        df.loc[i, 'id'] = use_id

df.dropna(subset = ['id'], inplace=True)
df.reset_index(drop=True, inplace=True)
df['id'] = df['id'].astype(int)
df
```

```python
# Создание столбцов для загружаемых метрик

years = [2021, 2022, 2023, 2024, 2025]
metrics = ['profit', 'revenue', 'EBIT']

for year in years:
    for metric in metrics:
        df[f'{year}_{metric}'] = np.nan
```

```python
# Парсинг нужных метрик

years = set([2021, 2022, 2023, 2024, 2025])
metrics = ['profit', 'revenue', 'EBIT']
session = requests.Session()
headers = {"User-Agent": "...Chrome..."}
n = len(df)

for i in range(n):
    temp = df['id'][i]
    url = f'https://bo.nalog.gov.ru/nbo/organizations/{temp}/bfo/'
    bfo = session.get(url, headers = headers).json()
    for year in bfo:
        year_num = year['period']
        if 'current2400' in year['typeCorrections'][0]['correction']['financialResult']:
            profit = year['typeCorrections'][0]['correction']['financialResult']['current2400']
            df.loc[i, f'{year_num}_profit'] = profit
        if 'current2110' in year['typeCorrections'][0]['correction']['financialResult']:
            revenue = year['typeCorrections'][0]['correction']['financialResult']['current2110']
            df.loc[i, f'{year_num}_revenue'] = revenue
        if 'current2200' in year['typeCorrections'][0]['correction']['financialResult']:
            ebit = year['typeCorrections'][0]['correction']['financialResult']['current2200']
            df.loc[i, f'{year_num}_EBIT'] = ebit
```

```python
# Расчёт изменения процента чистой прибыли за N = 4 года. Ещё тут отбрасывается процент больше 10 в любую из сторон
# (лучше, наверное, было оставить, но так график не очень красивый получится из-за малой выборки)

df['percent'] = (df['2025_profit'] - df['2021_profit']) / df['2021_profit']
data_percent = df['percent'].copy()
k = len(data_percent)
data_percent = data_percent[abs(data_percent) < 10]
m = len(data_percent)
data_percent.dropna(inplace=True)
data_percent.reset_index(drop=True, inplace=True)

print(1 - m / k) # Процент предприятий, чистая прибыль которых упала \ поднялась более чем на 10% за 4 года
```

```python
# Рисуем график

plt.hist(data_percent, bins=10, zorder=2)
plt.grid(True, alpha=0.5)
plt.xlabel('Процент роста чистой прибыли за 4 года, %')
plt.ylabel('Число компаний')
plt.title('Гистограмма распределения роста чистой прибыли за 4 года')
```

```python
# Забыл загрузить координаты. Загружаем

place_df = df.copy()
session = requests.Session()
headers = {"User-Agent": "...Chrome..."}
n = len(df)


place_df['latitude'] = np.nan
place_df['longitude'] = np.nan
for i in range(n):
    temp = place_df['id'][i]
    url = f'https://bo.nalog.gov.ru/nbo/organizations/{temp}'
    temp = session.get(url, headers = headers).json()
    if 'location' in temp:
        temp = temp['location']
        if temp:
            if 'latitude' in temp:
                place_df.loc[i, 'latitude']=temp['latitude']
            if 'longitude' in temp:
                place_df.loc[i, 'longitude']=temp['longitude']

```

```python
# Фильтрация по значению nan в ширине, долготе и прибыли в 2025-м. После этого оставляем 50 компаний с наибольшей прибылью
# В конце рисуем карту и отмечаем на ней полученные компании

place_df.dropna(subset=['longitude', 'latitude', '2025_revenue'], inplace=True)
place_df.reset_index(inplace=True, drop=True)
place_df.sort_values(by='2025_revenue', inplace=True)
place_df = place_df.tail(50)
place_df


fig = px.scatter_map(place_df, lat="latitude", lon="longitude", hover_name="Наименование / ФИО", zoom=3, height=300)
fig.update_layout(map_style="open-street-map")
fig.update_layout(margin={"r":0,"t":0,"l":0,"b":0})
fig.show()
```
